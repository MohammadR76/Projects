```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values=np.nan, strategy="mean")
imputer.fit(X[:,1:])
X[:,1:3] = imputer.transform(X[:,1:])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, LabelEncoder, StandardScaler
```

```python
le = LabelEncoder()
X[:, 0] = le.fit_transform(X[:,0])
print(X)
```

```python
ct = ColumnTransformer(transformers=[("Country", OneHotEncoder(), [0])], remainder="passthrough")
X = ct.fit_transform(X)
print(X)
```

```python
data["Region"].unique()
```

```python
Sta_Scal = StandardScaler()
X = Sta_Scal.fit_transform(X)
print(X)
```

```python
# Save Label_Encoder Model
with open("Label_Encoder.pkl", "wb") as f:
    pickle.dump(le, f)
```

```python
# Save Column_Transformer Model
with open("Column_Transformer.pkl", "wb") as f:
    pickle.dump(ct, f)
```

```python
# Save Standard_Scalar Model
with open("Standard_Scalar.pkl", "wb") as f:
    pickle.dump(Sta_Scal, f)
```
