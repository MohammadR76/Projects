# Fill Missing data

```python
data["Age"].fillna(data["Age"].mean(), inplace=True)
data
```

```python
data.dropna(inplace=True, subset=['Salary'])
data
```

#  Encoding categorical data

```python
X = data.iloc[:, :-1]
Y = data.iloc[:, -1]
```

```python
X
```

```python
X = pd.get_dummies(X, dtype=int)
X
```

```python
Y = (Y == "Yes").astype(int)
```

```python
Y
```

# Splitting the dataset into the Training set and Test set

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2)
```

```python
print(f"Shape of Train set: {X_train.shape}")
print(f"Shape of Test set: {X_test.shape}")
```

# Feature Scaling

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
sc = StandardScaler()
mm = MinMaxScaler()
```

```python
X_train
```

```python
X_train.describe()
```

```python
sc_X_train = sc.fit_transform(X_train)
sc_X_train = pd.DataFrame(sc_X_train, columns=X_train.columns, index=X_train.index)
sc_X_train
```

```python
sc_X_train.describe()
```

```python
# Example
num = X_train.iloc[0, 0]
mean = X_train["Age"].mean()
std = X_train["Age"].std()
norm_num = (num - mean) / std
print(f"Original: {num}, Normalized: {norm_num}")
```

```python
mm_X_train = mm.fit_transform(X_train)
mm_X_train = pd.DataFrame(mm_X_train, columns=X_train.columns, index=X_train.index)
mm_X_train
```

```python
mm_X_train.describe()
```

```python
# Example
num = X_train.iloc[0,0]
max = X_train["Age"].max()
min = X_train["Age"].min()
norm_num = (num - min) / (max - min)
print(f"Original: {num}, Normalized: {norm_num}")
```
