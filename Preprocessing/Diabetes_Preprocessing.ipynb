```python
data_nan = data.replace({"test": 0, "skin": 0}, np.nan, inplace=False)
data_nan
```

```python
from sklearn.impute import KNNImputer
from sklearn.preprocessing import MinMaxScaler, StandardScaler, Binarizer, Normalizer
```

```python
knn_imputer = KNNImputer(n_neighbors=3)
data_knn = knn_imputer.fit_transform(data_nan)
```

```python
data_knn_df = pd.DataFrame(data_knn, columns=data.columns)
data_knn_df.describe()
```

```python
data["skin"].hist(bins = 30, figsize = (10, 5), color="blue")
data_knn_df["skin"].hist(bins = 30, figsize = (10, 5), color="red")
```

```python
data["test"].hist(bins = 30, figsize = (10, 5), color="blue")
data_knn_df["test"].hist(bins = 30, figsize = (10, 5), color="red")
```

```python
scaler = MinMaxScaler(feature_range=(0, 1))
data_scaled = scaler.fit_transform(data_knn_df)
data_scaled_df = pd.DataFrame(data_scaled, columns=data.columns)
data_scaled_df
```

```python
data_scaled_df.describe()
```

```python
stscaler = StandardScaler()
data_stscaled = stscaler.fit_transform(data_knn_df)
data_stscaled_df = pd.DataFrame(data_stscaled, columns=data.columns)
data_stscaled_df
```

```python
normscaler = Normalizer()
data_normscaled = normscaler.fit_transform(data_knn_df)
data_normscaled = pd.DataFrame(data_normscaled, columns=data.columns)
data_normscaled
```

```python
binarize = Binarizer(threshold = 0.0)
data_binarize = binarize.fit_transform(data_stscaled_df)
data_binarize = pd.DataFrame(data_binarize, columns=data.columns)
data_binarize
```

```python
data_binarize.describe()
```

```python
plt.figure(figsize=(10,5))

data_scaled_df["test"].hist(bins=30, alpha=0.8, color="blue", label="Scaled")
data_stscaled_df["test"].hist(bins=30, alpha=0.5, color="red", label="Standardized")
data_normscaled["test"].hist(bins=30, alpha=0.6, color="green", label="Normalized")
data_binarize["test"].hist(bins=2, alpha=0.3, color="orange", label="Binarized")

plt.legend()
plt.show()
```
