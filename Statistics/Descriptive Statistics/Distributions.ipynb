```python
from scipy import stats
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
data_norm = stats.norm.rvs(size = 10000, loc = 0 , scale = 1)
```

```python
sns.displot(data_norm, bins = 100, kde = False)
```

```python
data_uniform = stats.uniform.rvs(size = 10000, loc = 0 , scale = 1)
```

```python
sns.displot(data_uniform, bins = 100, kde = False)
```

```python
data_bernouli = stats.bernoulli.rvs(size = 10000, p = 0.2)
```

```python
sns.displot(data_bernouli, kde = False)
```

```python
data_binom = stats.binom.rvs(size = 1000, n = 100, p = 0.2)
print(data_binom)
```

```python
sns.displot(data_binom, bins = 100, kde = False )
```

```python
p = 1/6
data_multinom = np.random.multinomial(n = 6, pvals=[p, p, p, p, p, p], size= 1000)
print(data_multinom)
```

```python
sns.displot(data_multinom, bins = 100, kde = False)
```

```python
icu = 0.3
surgery = 0.2
el = 0.5
data_multinomial = np.random.multinomial(n=1000, pvals=[icu, surgery, el], size=365)
print(data_multinomial)
```

```python
sns.displot(data_multinomial, bins = 100, kde = False)
```

```python
data_poison = np.random.poisson(lam=7, size=1000)
print(data_poison)
```

```python
sns.displot(data_poison, kde = False)
```

```python
data_exponential = np.random.exponential(scale=2, size=1000)
print(data_exponential)
```

```python
sns.displot(data_exponential)
```
