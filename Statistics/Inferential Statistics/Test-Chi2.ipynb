```python
ct = pd.crosstab(data["Attrition"], data["JobSatisfaction"], margins=True)
ct
```

```python
obs = np.append(ct.iloc[0][: -1].values, ct.iloc[1][: -1].values)
obs
```

```python
no = ct.iloc[0]/ct.iloc[0,-1]
no
```

```python
yes = ct.iloc[1]/ct.iloc[1,-1]
yes
```

```python
diff = no - yes
diff
```

```python
row_sum = ct.iloc[:2,-1].values
row_sum
```

```python
total = row_sum.sum()
total
```

```python
exp = []
for j in range(2):
  for val in ct.iloc[2, : -1].values:
    print(val, "*", row_sum[j], "/", total, "=", val*row_sum[j]/total)
    exp.append(val*row_sum[j]/total)
```

```python
chi2_formula = ((obs - exp)**2 / exp).sum()
chi2_formula
```

```python
dof = (len(row_sum)- 1)*(len(ct.iloc[2,:-1])-1)
dof
```

```python
# Table: alpha = 0.05, df = 3
chi2_table = 7.815

if chi2_formula > chi2_table:
    print("Reject H0")
else:
    print("Accept H0")
```

```python
chi2_cdf = 1 - stats.chi2.cdf(chi2_formula, dof)

if chi2_cdf < 0.05:
  print("Reject H0")
```

Based on the Chi-square test results (p < 0.05), there is a statistically significant relationship between job satisfaction and attrition.
