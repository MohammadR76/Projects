```python
from scipy.stats import shapiro, probplot
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
import pylab
warnings.filterwarnings("ignore")
```

```python
data = np.random.normal(loc = 20, scale = 5, size = 1000)
plt.hist(data, bins=100)
plt.show()
```

```python
data_df = pd.DataFrame(data, columns=["Number"])
data_df.describe()
```

```python
W, p_value = shapiro(data)

print("W-statistic:", W)
print("p-value:", p_value)
```

```python
if p_value > 0.05:
    print("The data is normally distributed.")
```

```python
probplot(data, dist="norm", plot=pylab)
pylab.title("Q-Q Plot")
pylab.show()
```
