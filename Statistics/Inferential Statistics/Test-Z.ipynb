```python
from scipy import stats
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
X_bar = 102
mu = 100
sigma = 16
n = 100
```

```python
z = (X_bar - mu) / (sigma / math.sqrt(n))
print(f"Z Value: {z}")
```

```python
plt.style.use('seaborn-v0_8')
plt.fill_between(
    x=np.arange(-4, 1.25, 0.01),
    y1=stats.norm.pdf(np.arange(-4, 1.25, 0.01)),
    facecolor='green'
)
plt.fill_between(
    x=np.arange(1.25, 4, 0.01),
    y1=stats.norm.pdf(np.arange(1.25, 4, 0.01)),
    facecolor='red'
)
```

image.png

```python
cdf_zero = stats.norm.cdf(0)
```

```python
cdf_z = stats.norm.cdf(z)
cdf_z
```

```python
# Area under curve between 0 and 1.25
area = cdf_z - cdf_zero
print(f"Area : {round(area,4)}")
```

```python
# Red Area in plot
red_area = 0.5 - area
print(f"Red Area : {round(red_area,4)}")
```

```python
print(f"P_value: {red_area}")
```
