```python
from scipy import stats
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from warnings import filterwarnings
filterwarnings('ignore')
```

```python
g1 = [20, 23, 21, 25, 18, 17, 18, 24, 20, 24, 23, 19]
g2 = [24, 25, 21, 22, 23, 28, 24, 27, 21, 23]
```

```python
g1_df = pd.DataFrame(g1, columns=["number"])
g2_df = pd.DataFrame(g2, columns=["number"])
```

```python
print("Describe of g1")
g1_df.describe()
```

```python
print("Describe of g2")
g2_df.describe()
```

```python
g1_df["group"] = "g1"
g2_df["group"] = "g2"
g2_df
```

```python
# Merge two Data Frame
merge_df = g1_df.merge(g2_df, on = ["number", 'group'], how="outer" )
```

```python
merge_df.sort_values("number").reset_index(drop=True)
merge_df.index = merge_df.index + 1
merge_df
```

```python
merge_df["rank"] = stats.rankdata(merge_df["number"], method="average")
merge_df
```

```python
R1 = merge_df.loc[merge_df['group'] == "g1"]["rank"].to_numpy().sum()
R2 = merge_df.loc[merge_df['group'] == "g2"]["rank"].to_numpy().sum()
n1 = len(g1)
n2 = len(g2)

print("R1: ", R1)
print("R2: ", R2)
print("n1: ", n1)
print("n2: ", n2)
```

```python
U1 = R1 - (n1*(n1+1)/2)
U2 = R2 - (n2*(n2+1)/2)
Uf = min(U1, U2)

print("U1:", U1)
print("U2:", U2)
print("U (final):", Uf)
```

```python
print(U1 + U2 == n1 * n2)
```

```python
mean = (n1 * n2) / 2
print("Mean: ", mean)

std = np.sqrt(n1 * n2 * (n1 + n2 + 1) / 12)
print("Std: ", std)
```

```python
Z_score = (Uf - mean) / std
print("Z_score: ", Z_score)
```

```python
# Two Sided
Z_score_table = 0.01876*2
print(f"p_value: {Z_score_table}")
```

```python
U_formula, p_value = stats.mannwhitneyu(g1, g2, alternative="two-sided")
print("U_formula: ", U_formula)
print("p_value: ", p_value)
```

```python
# g1 greater than g2
Z_score_table = 1 - 0.01876
print(f"p_value: {Z_score_table}")
```

```python
U_formula, p_value = stats.mannwhitneyu(g1, g2, alternative="greater")
print("U_formula: ", U_formula)
print("p_value: ", p_value)
```

```python
# g1 less than g2
Z_score_table =  0.01876
print(f"p_value: {Z_score_table}")
```

```python
U_formula, p_value = stats.mannwhitneyu(g1, g2, alternative="less")
print("U_formula: ", U_formula)
print("p_value: ", p_value)
```
